```python
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_community.document_loaders import WebBaseLoader
os.environ['LANGCHAIN_API_KEY'] = os.getenv('LANGCHAIN_API_KEY')
os.environ["LANGCHAIN_TRACING_V2"] = "true"


##load the document
data=WebBaseLoader("https://docs.langchain.com/oss/python/deepagents/overview")
lodded_data=data.load()
lodded_data
```

```python
## splittng the document
from langchain_text_splitters import RecursiveCharacterTextSplitter
Splitter=RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=10)
doc=Splitter.split_documents(lodded_data);
```

```python
##embedding the document
from langchain_community.embeddings import OllamaEmbeddings
from langchain_community.vectorstores import FAISS
embedding=OllamaEmbeddings(model="nomic-embed-text")
```

```python
##vectorStore
vectorStore=FAISS.from_documents(doc, embedding)
vectorStore
```

```python
### retrival from th vector tore db
response=vectorStore.similarity_search("What is LangChain DeepAgents?")
print(response)
```

```python
###now let retrieval it from the llm
retriver=vectorStore.as_retriever()
doc=retriver.invoke("what is langchain")
```

```python
##promptTemplate
from langchain_core.prompts import ChatPromptTemplate

prompt=ChatPromptTemplate(
    {
    """ 
    Answer the question as you are Ai expert from this context {context}:
    {question}
    """
    }
)
context = "Agents can use tools to take actions autonomously.\n\nLangChain is a framework for building LLM applications."

prompt.invoke({"context":{context},"question":"what is langchain"})
```

```python
from langchain_ollama import ChatOllama

llm=ChatOllama(model="llama3.1")

final=retriver|prompt|llm
final.invoke("what is langchain")
```
