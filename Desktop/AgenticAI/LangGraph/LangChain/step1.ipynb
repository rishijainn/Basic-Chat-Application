```python
import os
from dotenv import load_dotenv
load_dotenv()
from langchain_ollama import ChatOllama
os.environ['LANGCHAIN_API_KEY'] = os.getenv('LANGCHAIN_API_KEY')
os.environ["LANGCHAIN_TRACING_V2"] = "true"

llm=ChatOllama(model="llama3.1", temperature=0.9, max_tokens=3000)
result=llm.invoke("Write a poem about the ocean in the style of Shakespeare.")
print(result)
```
